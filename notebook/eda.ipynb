```python
missing_pct = dataset.isnull().mean() * 100
drop_cols = missing_pct[missing_pct >= 90].index.tolist()
print("dropping:", drop_cols)
dataset = dataset.drop(columns=drop_cols)
```

```python
# convert the column from text to numbers (anything that isn't a number becomes NaN)
age = pd.to_numeric(dataset["customerAge"], errors="coerce")

# find the impossible ages: younger than 18 OR older than 100
bad_age = (age < 18) | (age > 100)
print("impossible ages:", bad_age.sum())

# replace those impossible values with NaN (= "unknown"), but KEEP the row
dataset.loc[bad_age, "customerAge"] = np.nan
```

```python
# clean the amount text (remove £ and commas) and convert to numbers
amt = pd.to_numeric(dataset["amount"].str.replace(r"[£,]", "", regex=True), errors="coerce")

q1 = amt.quantile(0.25)          # 25% mark
q3 = amt.quantile(0.75)          # 75% mark
iqr = q3 - q1                    # spread of the middle half
lower = q1 - 1.5 * iqr           # lower fence
upper = q3 + 1.5 * iqr           # upper fence

is_outlier = (amt < lower) | (amt > upper)
print(f"normal range: {lower:.2f} to {upper:.2f}")
print(f"outliers found: {is_outlier.sum():,}")
```

```python
z = (amt - amt.mean()) / amt.std()     # how many standard deviations from the average
outliers_z = z.abs() > 3               # more than 3 away (either direction)
print("z-score outliers:", outliers_z.sum())
```

```python
cap = amt.quantile(0.999)        # the value 99.9% of amounts are below
amt_capped = amt.clip(upper=cap) # anything above the cap is set to the cap
```

```python
REGION_MAP = {r: r for r in ["london","south east","north west","east of england","west midlands",
    "east midlands","south west","north east","scotland","wales","northern ireland","overseas"]}
REGION_MAP.update({"gtr london":"london","greater london":"london","ldn":"london",
    "south-east":"south east","se england":"south east","north-west":"north west","n west":"north west",
    "east anglia":"east of england","eastern":"east of england","w midlands":"west midlands","w. midlands":"west midlands",
    "e midlands":"east midlands","e. midlands":"east midlands","south-west":"south west","sw england":"south west",
    "yorkshire and the humber":"yorkshire","yorkshire & humber":"yorkshire","yorks":"yorkshire",
    "north-east":"north east","n east":"north east","alba":"scotland","cymru":"wales",
    "ni":"northern ireland","n ireland":"northern ireland","n. ireland":"northern ireland"})

CATEGORY_MAPS = {
    "merchant_category": {"groceries":"groceries","grocery":"groceries","supermarkets":"groceries",
        "online_shopping":"online_shopping","online shopping":"online_shopping","online":"online_shopping","e-commerce":"online_shopping",
        "restaurants":"restaurants","dining":"restaurants","food & drink":"restaurants",
        "transport":"transport","travel_local":"transport","fuel":"fuel","petrol":"fuel","fuel/petrol":"fuel",
        "entertainment":"entertainment","leisure":"entertainment",
        "retail_clothing":"retail_clothing","clothing":"retail_clothing","retail - clothing":"retail_clothing","retail clothing":"retail_clothing",
        "cash_withdrawal":"cash_withdrawal","cash withdrawal":"cash_withdrawal","atm":"cash_withdrawal","cash":"cash_withdrawal",
        "utilities_bills":"utilities_bills","utilities/bills":"utilities_bills","bills":"utilities_bills","utilities":"utilities_bills",
        "electronics":"electronics","electrical":"electronics","travel":"travel","airlines/hotels":"travel",
        "gambling":"gambling","betting":"gambling"},
    "channel": {"contactless":"contactless","contact_less":"contactless","tap":"contactless",
        "chip_pin":"chip_pin","chip":"chip_pin","chip & pin":"chip_pin","chip and pin":"chip_pin",
        "online":"online","ecom":"online","cnp":"online","e-commerce":"online",
        "mobile_wallet":"mobile_wallet","mobile wallet":"mobile_wallet","apple_pay":"mobile_wallet","google_pay":"mobile_wallet",
        "magstripe":"magstripe","swipe":"magstripe"},
    "device_type": {"ios":"ios","apple":"ios","iphone":"ios","android":"android","desktop":"desktop","pc":"desktop","windows":"desktop"},
    "card_type": {"visa":"visa","mastercard":"mastercard","mc":"mastercard","amex":"amex","american express":"amex"},
    "customer_home_region": REGION_MAP,
    "merchant_region": REGION_MAP,
}
CURRENCY_MAP = {"gbp":"GBP","£":"GBP","ukp":"GBP","eur":"EUR","€":"EUR","usd":"USD","$":"USD","try":"TRY"}
TO_GBP = {"GBP":1.0,"EUR":0.85,"USD":0.79,"TRY":0.024}
YES_NO = {"true":1,"y":1,"1":1,"false":0,"n":0,"0":0}

# drop leaky/junk columns, bad rows, duplicate transaction ids
df = dataset.drop(columns=["chargeback_filed","bank_code","card_last4","merchant_name"])
df = df[~df["customer_id"].str.startswith("TEST", na=False)].dropna(subset=["is_fraud"]).drop_duplicates(subset="transaction_id")
df["is_fraud"] = df["is_fraud"].astype(int)
df["customer_id"] = df["customer_id"].str.extract(r"(\d+)", expand=False).astype(int)

# timestamps: try UK format, then US, then ISO, keep whichever parses
ts = df["Transaction Timestamp"]
df["timestamp"] = (pd.to_datetime(ts, format="%d/%m/%Y %H:%M", errors="coerce")
    .fillna(pd.to_datetime(ts, format="%m/%d/%Y %I:%M %p", errors="coerce"))
    .fillna(pd.to_datetime(ts.str.replace("T"," ").str.replace("Z",""), format="%Y-%m-%d %H:%M:%S", errors="coerce")))
df = df[df["timestamp"].dt.year == 2024].drop(columns=["Transaction Timestamp"])

# amounts: strip £/commas, drop placeholders, convert everything to GBP
df["amount"] = pd.to_numeric(df["amount"].str.replace("£","").str.replace(",",""), errors="coerce")
df = df[(df["amount"] > 0) & (df["amount"] < 999999)]
df["currency"] = df["currency"].str.lower().map(CURRENCY_MAP).fillna("GBP")
df["amount_gbp"] = (df["amount"] * df["currency"].map(TO_GBP)).round(2)

# other numeric columns: impossible values -> NaN
df["customerAge"] = pd.to_numeric(df["customerAge"], errors="coerce").where(lambda s: s.between(18, 100))
df["customer_annual_income"] = pd.to_numeric(df["customer_annual_income"], errors="coerce").where(lambda s: s.between(1, 5_000_000))
dist = pd.to_numeric(df["distance_from_home_km"], errors="coerce").abs()
df["distance_from_home_km"] = dist.where(dist <= 20000, dist / 1000)   # metres -> km
df["failed_attempts_24h"] = pd.to_numeric(df["failed_attempts_24h"], errors="coerce").clip(lower=0)
df["credit_limit_gbp"] = pd.to_numeric(df["credit_limit_gbp"], errors="coerce")

# yes/no text -> 1/0
for col in ["card_present", "is_international"]:
    df[col] = df[col].str.lower().map(YES_NO)
df["auth_approved"] = df["auth_response"].str.lower().isin(["approved", "a"]).astype(int)
df = df.drop(columns=["auth_response"])

# one spelling per category
for col, mapping in CATEGORY_MAPS.items():
    df[col] = df[col].str.lower().map(mapping)
df["merchant_country"] = df["merchant_country"].str.upper().replace("UK", "GB")
for col in ["card_kind", "card_tier", "transaction_type", "source_system"]:
    df[col] = df[col].str.lower()

df = df.sort_values("timestamp").reset_index(drop=True)
print(df.shape)
print(df.isna().mean().round(3).sort_values(ascending=False).head(10))
df.to_csv("../data/processed/transactions_clean.csv", index=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.select_dtypes(include="number").drop(columns=["customer_id", "mcc"]).corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation heatmap")
plt.tight_layout()
plt.show()
```
